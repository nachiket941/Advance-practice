```python
# 7. Calculate electricity bill

units = int(input("Enter electricity units: "))

if units <= 100:
    bill = units * 3
elif units <= 200:
    bill = 100 * 3 + (units - 100) * 5
elif units <= 300:
    bill = 100 * 3 + 100 * 5 + (units - 200) * 7
else:
    bill = 100 * 3 + 100 * 5 + 100 * 7 + (units - 300) * 10

print("Total Electricity Bill =", bill)
```

```python
# 9. Find the type of triangle

a = int(input("Enter first side: "))
b = int(input("Enter second side: "))
c = int(input("Enter third side: "))

if a <= 0 or b <= 0 or c <= 0 or a + b <= c or a + c <= b or b + c <= a:
    print("Invalid Triangle")
elif a == b == c:
    print("Equilateral Triangle")
elif a == b or b == c or a == c:
    print("Isosceles Triangle")
else:
    print("Scalene Triangle")
```

```python
# 10. Create a simple ATM menu

balance = 5000

print("1. Withdraw")
print("2. Deposit")
print("3. Balance")

choice = int(input("Enter your choice: "))

if choice == 1:
    amount = int(input("Enter withdrawal amount: "))
    if 0 < amount <= balance:
        balance -= amount
        print("Withdrawal Successful")
    else:
        print("Invalid amount or insufficient balance")

elif choice == 2:
    amount = int(input("Enter deposit amount: "))
    if amount > 0:
        balance += amount
        print("Deposit Successful")
    else:
        print("Invalid amount")

elif choice == 3:
    print("Balance =", balance)

else:
    print("Invalid Choice")
```

```python
# 13. Print prime numbers between 1 and n

n = int(input("Enter n: "))

for num in range(2, n + 1):
    prime = True

    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            prime = False
            break

    if prime:
        print(num, end=" ")
```

```python
# 14. Find factorial using a loop

num = int(input("Enter a number: "))

factorial = 1

if num < 0:
    print("Factorial is not defined for negative numbers")
else:
    for i in range(1, num + 1):
        factorial *= i

    print("Factorial =", factorial)
```

```python
# 16. Check Armstrong number

num = int(input("Enter a number: "))

temp = num
power = len(str(num))
total = 0

while temp > 0:
    digit = temp % 10
    total += digit ** power
    temp //= 10

if num >= 0 and total == num:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

```python
# 17. Reverse a number and check palindrome

num = int(input("Enter a number: "))

temp = num
reverse = 0

while temp > 0:
    digit = temp % 10
    reverse = reverse * 10 + digit
    temp //= 10

print("Reverse =", reverse)

if num >= 0 and num == reverse:
    print("Palindrome Number")
else:
    print("Not a Palindrome")
```

```python
# 18. Find GCD of two numbers

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

x = abs(a)
y = abs(b)

while y != 0:
    x, y = y, x % y

print("GCD =", x)
```

```python
# 19. Check palindrome using function

def palindrome(text):
    if text == text[::-1]:
        return True
    else:
        return False

word = input("Enter a string: ")

if palindrome(word):
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# 20. Count vowels and consonants

def count_letters(text):
    vowels = 0
    consonants = 0

    for ch in text.lower():
        if ch.isalpha():
            if ch in "aeiou":
                vowels += 1
            else:
                consonants += 1

    print("Vowels =", vowels)
    print("Consonants =", consonants)

text = input("Enter a string: ")

count_letters(text)
```

```python
# 21. Calculate simple and compound interest

def interest(p, r, t):
    si = (p * r * t) / 100
    ci = p * (1 + r / 100) ** t - p

    print("Simple Interest =", si)
    print("Compound Interest =", ci)

p = float(input("Enter principal: "))
r = float(input("Enter rate: "))
t = float(input("Enter time in years: "))

interest(p, r, t)
```

```python
# 22. Find all factors of a number

def factors(num):
    result = []

    for i in range(1, num + 1):
        if num % i == 0:
            result.append(i)

    return result

num = int(input("Enter a number: "))

if num > 0:
    print("Factors =", factors(num))
else:
    print("Enter a positive number")
```

```python
# 23. Find second-largest number

def second_largest(numbers):
    unique = list(set(numbers))
    unique.sort()

    if len(unique) < 2:
        return "Second largest does not exist"

    return unique[-2]

numbers = [10, 20, 40, 30, 40, 50]

print("Second Largest =", second_largest(numbers))
```

```python
# 28. Find common elements between two lists

list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]

common = []

for num in list1:
    if num in list2 and num not in common:
        common.append(num)

print("Common Elements =", common)
```
